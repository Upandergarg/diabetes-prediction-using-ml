Data Cleaning

```python
zero_columns = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI"
]

(df[zero_columns] == 0).sum()
```

replace the zeros with NaN

earlier it has non-zero values in those columns, now we converted invalid-looking zeroes into NaN

```python
df[zero_columns] = df[zero_columns].replace(0, np.nan)
df.isnull().sum()
```

handling missing values

1. drop missing one
2. imputation-> replace with median

```python
for column in zero_columns:
    df[column] = df[column].fillna(df[column].median())

df.isnull().sum()
```

checking for duplicated rows

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
df.duplicated().sum()
```
